## Introduction
This project aims to create an agent to sink all ships in the Battleship game in as little moves as possible and compare it with an agent that makes purely random moves

## Game Implementation
We implemented a single-player version of the Battleship game as shown below.

Players can:
* create custom boards using `place_ships(placements)`
* take shots on untried tiles using `play(row, col)`
* view the state of the board after the played moves using `display_board()`

```python
class Battleship:

    def __init__(self, rows=10, columns=10):
        self.rows = rows
        self.columns = columns
        self.moves = 0

        self.board = [
            [0 for _ in range(columns)]
            for _ in range(rows)
        ]

        self.exposed_board = [
            ["-" for _ in range(columns)]
            for _ in range(rows)
        ]

        # store tiles belonging to each ship
        self.ship_sizes = {}

        # store number of tiles hit so far
        self.ship_hits = {}

    # TODO: implement random placement generator
    def place_ships(self, placements):
        # reset everything first
        self.board = [
            [0 for _ in range(self.columns)]
            for _ in range(self.rows)
        ]

        self.exposed_board = [
            ["-" for _ in range(self.columns)]
            for _ in range(self.rows)
        ]

        self.ship_sizes = {}
        self.ship_hits = {}
        self.moves = 0

        for ship_id, ship in enumerate(placements, start=1):

            if not ship:
                raise ValueError("A ship cannot be empty.")

            # validation
            # TODO: check x or y-axis is the same for all tiles of same ship
            for row, col in ship:
                if not (
                    0 <= row < self.rows
                    and 0 <= col < self.columns
                ):
                    raise ValueError(
                        f"Coordinate out of bounds: {(row, col)}"
                    )

                if self.board[row][col] != 0:
                    raise ValueError(
                        f"Overlapping ship at {(row, col)}"
                    )

            # Place the ship using its index as its ID
            self.ship_sizes[ship_id] = len(ship)
            self.ship_hits[ship_id] = 0

            for row, col in ship:
                self.board[row][col] = ship_id

    def is_valid_move(self, row, col):
        return (
            0 <= row < self.rows
            and 0 <= col < self.columns
            and self.exposed_board[row][col] == "-"
        )

    def play(self, row, col):
        if not self.is_valid_move(row, col):
            print(f"Invalid move: ({row}, {col})")
            return None

        self.moves += 1

        ship_id = self.board[row][col]

        if ship_id == 0:
            self.exposed_board[row][col] = "0"
            print("Miss!")
            return (False, False, [])

        self.ship_hits[ship_id] += 1

        # check if ship is sunk
        sunk = (
            self.ship_hits[ship_id]
            == self.ship_sizes[ship_id]
        )

        sunk_tiles = []        

        if sunk:
            for r in range(self.rows):
                for c in range(self.columns):
                    if self.board[r][c] == ship_id:
                        sunk_tiles.append((r,c))
                        # reveal all tiles belonging to sunken ship
                        self.exposed_board[r][c] = str(ship_id)

            print(f"Hit! Ship {ship_id} sunk!")
        else:
            self.exposed_board[row][col] = "X"
            print(f"Hit!")

        return (True, sunk, sunk_tiles)

    
    def display_board(self, show_ships=False):
        board = self.board if show_ships else self.exposed_board

        print("   " + "".join(f"{c:3}" for c in range(self.columns)))

        print("   +" + "---" * self.columns + "+")

        for r in range(self.rows):
            print(f"{r:2} | " + "".join(
                f"{str(cell):3}" for cell in board[r]
            ) + "|")


    def is_game_over(self):
        return (
            bool(self.ship_sizes)
            and all(
                self.ship_hits[ship_id] == size
                for ship_id, size in self.ship_sizes.items()
            )
        )
```

## Random Agent
We created an agent that randomly chooses moves from the remaining unplayed moves to benchmark against

```python
import random

class RandomAgent:

    def play(self, board: Battleship):
        moves = 0

        while not board.is_game_over():
            moves = moves+ 1
            valid_moves = [
                (row, col)
                for row in range(board.rows)
                for col in range(board.columns)
                if board.is_valid_move(row, col)
            ]

            row, col = random.choice(valid_moves)
            board.play(row, col)

        print(f"Game won with {board.moves} moves")
        board.display_board()
        return moves
```

## Search Agent

We created an agent that decides the next possible move based on whether the previous move is a hit or miss.

1. The agent makes its first move in the centre of the board where the probabiltiy of a ship is higher.
2. If miss, the agent searches in a checkerboard pattern to halve the search space as the minimum length of a ship following standard Battleship rules is 2.
3. If hit, the agent searches adjacent tiles to find the remainder of the ship, and resume checkerboard search once sunk.
4. If there are no more moves continuining from the above, randomly choose a tile to continue from.

```python
from queue import Queue
import random

class SearchAgent:

    def __init__(self):
        self.dia_dirs = [
            (1, 1),
            (1, -1),
            (-1, 1),
            (-1, -1)
        ]

        self.adj_dirs = [
            (0, 1),
            (1, 0),
            (0, -1),
            (-1, 0)
        ]

    def play(self, board: Battleship):
        moves = 0
        self.visited = set()
        self.queue = Queue()
        self.stack = []

        row = board.rows // 2
        col = board.columns // 2

        while not board.is_game_over():
            moves += 1
            print(f"Move {moves}: {row, col}")

            is_hit, is_sunk, sunk_tiles = board.play(row, col)

            self.visited.add((row, col))

            if is_sunk:
                self.handle_sunk(sunk_tiles)

            elif is_hit:
                self.adjacent_search(board, row, col)
            else:
                self.diagonal_search(board, row, col)

            next_move = self.next_move()

            if next_move is not None:
                row, col = next_move
            else:
                row, col = self.rand_move(board)

        print(f"Game won with {moves} moves")
        board.display_board()
        return moves

    def next_move(self):
        while self.stack:
            move = self.stack.pop()
            if move not in self.visited:
                return move

        while not self.queue.empty():
            move = self.queue.get()
            if move not in self.visited:
                return move

    def adjacent_search(self, board, row, col):
        for dir in self.adj_dirs:
            new_row, new_col = row + dir[0], col + dir[1]
            if board.is_valid_move(new_row, new_col):
                self.stack.append((new_row, new_col))

    def diagonal_search(self, board, row, col):
        for dir in self.dia_dirs:
            new_row, new_col = row + dir[0], col + dir[1]
            if board.is_valid_move(new_row, new_col):
                self.queue.put((new_row, new_col))

    def handle_sunk(self, sunk_tiles):
        sunk_neighbours = set()

        for row, col in sunk_tiles:
            for dr, dc in self.adj_dirs:
                nr, nc = row + dr, col + dc
                sunk_neighbours.add((nr, nc))

        self.stack = [
            move for move in self.stack
            if move not in sunk_neighbours
        ]

    def rand_move(self, board):
        valid_moves = [
            (row, col)
            for row in range(board.rows)
            for col in range(board.columns)
            if board.is_valid_move(row, col)
        ]
        
        row, col = random.choice(valid_moves)
        return row, col


# Example Run
board.place_ships(placements)
agent = SearchAgent()
agent.play(board)
```
